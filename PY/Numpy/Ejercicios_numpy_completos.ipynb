## Ejercicio 11

Los intervalos del enunciado se solapan en sus extremos. Para evitar ambigüedades se da prioridad al intervalo inferior: 3 → 0, 5 → 4, 6 → 5, 7 → 6 y 9 → 8.

```python
def f(numero):
    if numero <= 3:
        return 0
    elif numero <= 5:
        return 4
    elif numero <= 6:
        return 5
    elif numero <= 7:
        return 6
    elif numero <= 9:
        return 8
    else:
        return 10

a = np.array([1, 5, 4, 7, 3, 9, 8, 6])
resultado = np.array([f(numero) for numero in a])
print(resultado)
```

## Ejercicio 15

Las conversiones fuera del rango de 8 bits se realizan con módulo 256 para ilustrar el resultado sin depender del comportamiento de conversión de versiones concretas de NumPy.

Enteros [1, -2, 3] a uint8

```python
a = np.array([1, -2, 3])
a = (a % 256).astype(np.uint8)
print(a, a.nbytes, "bytes")
```

[1, 254, 255] a uint8

```python
b = np.array([1, 254, 255])
b = b.astype(np.uint8)
print(b, b.nbytes, "bytes")
```

[1, 127, 128] a int8

```python
b = np.array([1, 127, 128])
b = ((b + 128) % 256 - 128).astype(np.int8)
print(b, b.nbytes, "bytes")
```

[1, -128, -129] a int8

```python
b = np.array([1, -128, -129])
b = ((b + 128) % 256 - 128).astype(np.int8)
print(b, b.nbytes, "bytes")
```

[1, 2, 3] a float32

```python
b = np.array([1, 2, 3])
b = b.astype(np.float32)
print(b, b.nbytes, "bytes")
```

[1.9, 2.01, 3.51] a int32

```python
b = np.array([1.9, 2.01, 3.51])
b = b.astype(np.int32)
print(b, b.nbytes, "bytes")
```

## Ejercicio 17

En el código original, `x` era una lista y no permitía operaciones como `1 - x`. Se convierte a un array float. Cada gráfica se muestra con Matplotlib.

```python
import matplotlib.pyplot as plt

def dibujar(x):
    y = 3*(1-x)**2 * np.exp(-x**2) - 10*(x/5-x**3)*np.exp(-x**2) - 1./3*np.exp(-(x+1)**2)
    figure = plt.figure(figsize=(8, 5))
    axes = figure.add_subplot()
    axes.plot(x, y)
    axes.set_title(f"Gráfica con {len(x)} valores")
    axes.grid(True)
    plt.show()
```

```python
x = np.array([-3, -2, -1, 0, 1, 2, 3], dtype=float)
dibujar(x)
```

```python
x = np.linspace(-3, 3, 10)
dibujar(x)
```

```python
x = np.linspace(-3, 3, 20)
dibujar(x)
```

## Ejercicio 18

El archivo `wine.csv` no se adjuntó. Esta celda carga el fichero cuando esté en la misma carpeta del cuaderno. Se asume que la última columna contiene el tipo de vino y las demás son características numéricas, sin encabezado.

```python
def imprimir_datos(matriz):
    x = matriz[:, :-1]
    y = matriz[:, -1]
    print("Número de características por vino:", x.shape[1])
    print("Número de muestras de vino:", matriz.shape[0])
    tipos = np.unique(y)
    print("Número de tipos distintos:", len(tipos))
    print("Tipos de vino:", tipos)
    print("Máximos de cada característica:", np.max(x, axis=0))
    print("Mínimos de cada característica:", np.min(x, axis=0))
    return x, y

try:
    datos_vino = np.loadtxt("wine.csv", delimiter=",")
    x_vino, y_vino = imprimir_datos(datos_vino)
except OSError:
    print("Falta wine.csv. Coloca el archivo junto al cuaderno y vuelve a ejecutar esta celda.")
```

## Ejercicio 19

```python
np.random.seed(0)
sales_data = np.random.normal(500, 50, 365)
sales_data[[50, 120, 180, 250, 300]] = [1500, 1400, 1600, 1700, 1800]
```

Detectar y eliminar valores fuera de tres desviaciones estándar

```python
media = np.mean(sales_data)
desviacion = np.std(sales_data)
mascara = np.abs(sales_data - media) <= 3 * desviacion
ventas_limpias = sales_data[mascara]
dias_originales = np.arange(1, 366)[mascara]
print("Valores atípicos:", sales_data[~mascara])
print("Días atípicos:", np.arange(1, 366)[~mascara])
```

Estadísticos tras limpiar

```python
print("Promedio:", np.mean(ventas_limpias))
print("Mediana:", np.median(ventas_limpias))
print("Desviación estándar:", np.std(ventas_limpias))
```

Vista de las ventas de cada trimestre (365 días divididos en 90, 91, 92 y 92 días, aproximadamente según los meses de un año no bisiesto). Se conservan los días originales al excluir atípicos.

```python
limites = [(1, 90), (91, 181), (182, 273), (274, 365)]
for i, (inicio, fin) in enumerate(limites):
    trimestre = ventas_limpias[(dias_originales >= inicio) & (dias_originales <= fin)]
    print(f"Trimestre {i+1}:", trimestre)
```

Cinco días con las ventas más bajas, indicando el día original

```python
indices = np.argsort(ventas_limpias)[:5]
for i in indices:
    print("Día:", dias_originales[i], "Ventas:", ventas_limpias[i])
```

## Ejercicio 20

```python
np.random.seed(3)
ages = np.random.randint(18, 70, size=200)
income = np.random.normal(50000, 15000, size=200)
education_years = np.random.randint(0, 20, size=200)
```

Correlación entre años de educación e ingresos

```python
correlacion = np.corrcoef(education_years, income)[0, 1]
print("Correlación:", correlacion)
```

Ingresos promedio por grupos de edad

```python
grupos = [(18, 30), (31, 45), (46, 60), (61, 69)]
for inicio, fin in grupos:
    ingresos_grupo = income[(ages >= inicio) & (ages <= fin)]
    print(f"Edades {inicio}-{fin}: {np.mean(ingresos_grupo):.2f}")
```

Percentil 90 y número de personas por encima

```python
p90 = np.percentile(income, 90)
print("Percentil 90:", p90)
print("Personas por encima:", np.sum(income > p90))
```

Ingresos promedio según años de educación

```python
for anios in np.unique(education_years):
    ingresos_grupo = income[education_years == anios]
    print(f"{anios} años de educación: {np.mean(ingresos_grupo):.2f}")
```
