
33. Elimina los outliers dentro de la lista del ejercicio de ejemplo de Pandas.


```python
Q1 = serie.quantile(0.25)
Q3 = serie.quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

resultado = serie[
    (serie >= limite_inferior) &
    (serie <= limite_superior)
]

print("Serie original:")
print(serie)

print("Serie sin outliers:")
print(resultado)
```

34. En una Serie de Pandas donde se registran las ventas mensuales de un producto durante años, utiliza el método group by para calcular las ventas medias en cada mes.

```python
fechas = pd.date_range(
    start='2024-01-01',
    periods=24,
    freq='MS'
)

ventas = pd.Series(
    [100, 120, 150, 130, 140, 160,
     170, 180, 190, 200, 210, 220,
     110, 130, 160, 140, 150, 170,
     180, 190, 200, 210, 220, 230],
    index=fechas
)

resultado = ventas.groupby(ventas.index.month).mean()

resultado.index = [
    'Enero', 'Febrero', 'Marzo', 'Abril',
    'Mayo', 'Junio', 'Julio', 'Agosto',
    'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'
]

print("Ventas medias por mes:")
print(resultado)
```
